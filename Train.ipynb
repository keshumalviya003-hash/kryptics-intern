```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Train.csv")
```

```python
Q1 = df['Item_Outlet_Sales'].quantile(0.25)

Q3 = df['Item_Outlet_Sales'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

outliers = df[
    (df['Item_Outlet_Sales'] < lower) |
    (df['Item_Outlet_Sales'] > upper)
]

print("Total Outliers:", len(outliers))
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x='Outlet_Type',
    y='Item_Outlet_Sales'
)

plt.title('Sales Distribution by Outlet Type')
plt.xlabel('Outlet Type')
plt.ylabel('Item Outlet Sales')
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```


The boxplot shows that Supermarket Type3 has the highest median sales, while Grocery Store has the lowest median sales. Supermarket Type3 also shows a wider spread of sales values. All four outlet types contain potential outliers, indicating some unusually high sales values.

```python
plt.figure(figsize=(9,5))

sns.scatterplot(
    data=df,
    x='Item_MRP',
    y='Item_Outlet_Sales',
    alpha=0.5
)

plt.title('Relationship Between MRP and Sales')
plt.xlabel('Item MRP')
plt.ylabel('Item Outlet Sales')

plt.show()
```


The scatter plot shows a positive relationship between Item MRP and Item Outlet Sales. Products with higher MRP generally show higher sales values, although sales vary considerably at similar MRP levels. This suggests that MRP is positively associated with sales, but it is not the only factor affecting sales.

```python
numeric_df = df.select_dtypes(include='number')

correlation = numeric_df.corr()

plt.figure(figsize=(10,7))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Numerical Columns')

plt.show()
```

 Interpretation

The heatmap shows that Item_MRP has the strongest positive correlation (0.57) with Item_Outlet_Sales, indicating a moderate positive relationship between them. Item_Visibility has a weak negative correlation (-0.13) with sales, while Item_Weight (0.01) and Outlet_Establishment_Year (-0.05) show almost no linear relationship with sales.

 Key Finding

Item_MRP is the most strongly correlated numerical feature with Item_Outlet_Sales in this dataset.
